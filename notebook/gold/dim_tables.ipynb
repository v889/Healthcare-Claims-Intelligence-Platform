```python
from pyspark.sql.functions import current_timestamp

dim_patient = (
    spark.table(f"{env_catalog}.silver.patients")
    .withColumn("gold_load_timestamp", current_timestamp())
)
```

```python
dim_patient.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(f"{env_catalog}.gold.dim_patient")
```

```python
dim_provider = (
    spark.table(f"{env_catalog}.silver.providers")
    .withColumn("gold_load_timestamp", current_timestamp())
)

dim_provider.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(f"{env_catalog}.gold.dim_provider")
```
